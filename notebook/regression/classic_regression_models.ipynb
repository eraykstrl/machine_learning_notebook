# Polynomal Regression

```python
import pandas as pd
```

```python
df = pd.read_csv('../job_salary_prediction_dataset.csv')
```

```python
X = df.drop('salary',axis = 1)
y = df['salary']
```

```python
from sklearn.preprocessing import PolynomialFeatures

poly_reg_2 = PolynomialFeatures(degree=2)
poly_reg_3 = PolynomialFeatures(degree=3)
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(
  X,
  y,
  random_state=42,
  test_size=0.2
)
```

```python
numeric_col = X.select_dtypes(include=['number']).columns
```

```python
X_train_poly_2 = poly_reg_2.fit_transform(X_train[numeric_col])
X_test_poly_2 = poly_reg_2.transform(X_test[numeric_col])
```

```python
X_train_poly_3 = poly_reg_3.fit_transform(X_train[numeric_col])
X_test_poly_3 = poly_reg_3.transform(X_test[numeric_col])
```

```python
categoric_col = X.select_dtypes(include=['category','object']).columns
```

```python
from sklearn.preprocessing import OrdinalEncoder

ord_encoder = OrdinalEncoder()
```

```python
X_train_encoded = ord_encoder.fit_transform(X_train[categoric_col])
X_test_encoded = ord_encoder.transform(X_test[categoric_col])
```

```python
import numpy as np
```

```python
import numpy as np

X_train_poly_2_last = np.hstack([X_train_poly_2, X_train_encoded])
X_test_poly_2_last = np.hstack([X_test_poly_2, X_test_encoded])
```

```python
from sklearn.linear_model import LinearRegression

linear_reg = LinearRegression()

model = linear_reg.fit(X_train_poly_2_last,y_train)
```

```python
y_preds = model.predict(X_test_poly_2_last)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error

r2_score_l = r2_score(y_test,y_preds)
mae_score_l = mean_absolute_error(y_test,y_preds)
mse = mean_squared_error(y_test,y_preds)
```
