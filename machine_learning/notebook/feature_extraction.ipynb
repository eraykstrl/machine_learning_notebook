# Encoding String (Categoric, Object) Data

- **One-Hot Encoding:** Converts each category into separate binary columns, making no assumption about order.  
- **Label Encoding:** Assigns a unique integer to each category but may introduce unintended ordinal relationships.  
- **Ordinal Encoding:** Encodes categories with a defined order using integer values that preserve ranking.  
- **Target Encoding:** Replaces each category with the mean of the target variable for that category.  
- **Frequency Encoding:** Maps each category to its frequency (or count) in the dataset.  
- **Binary Encoding:** Converts categories into binary numbers to reduce dimensionality compared to one-hot encoding.  
- **Embedding:** Represents categories as dense numerical vectors learned during model training, commonly used in deep learning.

For can use these encoding methods firstly we find categoric data

## OneHotEncoding

```python
import pandas as pd

df = pd.read_csv('../job_salary_prediction_dataset.csv')
```

```python
df.head()
```

Don't apply these encoder without seperate X and y columns.
Don't apply these encoder without seperate train and test data.

```python
X = df.drop(["salary"],axis = 1)
y = df["salary"]
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(
  X,
  y,
  random_state=42,
  test_size=0.2
)
```

```python
categoric_X_train = X_train.select_dtypes(include=['category','object']).columns
categoric_X_test = X_test.select_dtypes(include=['category','object']).columns
```

```python
from sklearn.preprocessing import OneHotEncoder

one_hot_x = OneHotEncoder(handle_unknown="ignore", sparse_output=False)

encoded_train = pd.DataFrame(
    one_hot_x.fit_transform(X_train[categoric_X_train]),
    columns=one_hot_x.get_feature_names_out(categoric_X_train),
    index=X_train.index
)

encoded_test = pd.DataFrame(
    one_hot_x.transform(X_test[categoric_X_test]),   
    columns=one_hot_x.get_feature_names_out(categoric_X_test),
    index=X_test.index
)

X_train = X_train.drop(columns=categoric_X_train).join(encoded_train)
X_test  = X_test.drop(columns=categoric_X_test).join(encoded_test)
```

# Normalization

We have different normalization methods that can transform data into new structure.  
Algorithms can process these normalized data on the same scale.

1. **Standard Scaler**: It is known as z-normalization and uses mean and variance.
2. **Min-Max Scaler**: This method uses min and max values.
3. **Box-Cox Transformation**: The Box-Cox transformation uses a power transformation method. It reduces right skewness and makes the distribution closer to normal.
4. **Yeo-Johnson Transformation**: The Yeo–Johnson transformation can handle both positive and negative values. It reduces skewness and makes the distribution closer to normal.
