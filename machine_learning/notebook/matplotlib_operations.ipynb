```python
import pandas as pd

df = pd.read_csv('../job_salary_prediction_dataset.csv')
```

```python
df.head()
```

```python
df.describe()
```

```python
X = df.drop('salary',axis = 1)
```

```python
y = df['salary']
```

```python
%pip install scikit-learn
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(
  X,
  y,
  random_state=42,
)
```

```python
from sklearn.preprocessing import OrdinalEncoder

ordinal_encoder = OrdinalEncoder(handle_unknown="use_encoded_value",unknown_value=-1)

categoric_columns = X.select_dtypes(include=["object"]).columns


X_train[categoric_columns] = ordinal_encoder.fit_transform(X_train[categoric_columns])

X_test[categoric_columns] = ordinal_encoder.transform(X_test[categoric_columns])
```

```python
from sklearn.metrics import r2_score
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_absolute_percentage_error
from sklearn.linear_model import Ridge,Lasso,LinearRegression
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
import numpy as np



models = {
  "Ridge Lineer Regressor" : Ridge(),
  "Lasso Lineer Regressor" : Lasso(),
  "Linear Regression" : LinearRegression(),
  "Support Vector Regression" :SVR(),
  "Decision Tree Regression" : DecisionTreeRegressor()
}

performance_metrics = {

}


for name,model in models.items():

  model.fit(X_train,y_train)
  y_pred = model.predict(X_test)

  mae = mean_absolute_error(y_test,y_pred)
  mse = mean_squared_error(y_test,y_pred)
  rmse = np.sqrt(mse)
  r2_s = r2_score(y_test,y_pred)
  mape= mean_absolute_percentage_error(y_test,y_pred)

  performance_metrics[name] = {"mae" : mae,"mse" :mse,"rmse":rmse,"mape" : mape,"r2_score" : r2_s}


algorithms = performance_metrics.keys()

rmse_values = [metric["rmse"] for metric in performance_metrics.values()]
mse_values= [metric["mse"] for metric in performance_metrics.values()]
mae_values = [metric["mae"] for metric in performance_metrics.values()]
mape_values = [metric["mape"] for metric in performance_metrics.values()]
r2_values = [metric["r2_score"] for metric in performance_metrics.values()]


```

```python
import matplotlib.pyplot as plt

def plot_models_performance(metric_name,metric_value,color):

  plt.figure(figsize = (10,6))
  plt.bar(algorithms,metric_value,color = color)
  plt.xlabel("Algoritmalar")
  plt.ylabel(metric_name)
  plt.title("Some algorithm performance")
  plt.show()

plot_models_performance("rmse",rmse_values,color = "red")
plot_models_performance("mse",mse_values,color = "blue")
plot_models_performance("r2 value",r2_values,color = "green")
plot_models_performance("mae",mae_values,color = "orange")
plot_models_performance("mape",mape_values,color = "purple")



```
