# Regression is a method used to predict continous numerical values based on input variables.

```python
import pandas as pd

df = pd.read_csv('../job_salary_prediction_dataset.csv')
```

```python
df.head()
```

```python
df.tail()
```

```python
X = df.drop('salary',axis=1)
```

```python
X.isnull().sum()
```

```python
y = df['salary']
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(
  X,
  y,
  random_state=42,
  test_size=0.2
)
```

```python
X_train.head()
```

```python
from sklearn.impute import SimpleImputer
import numpy as np

imputer_X = SimpleImputer(missing_values=np.nan,strategy = "median")
```

```python
numeric_X = X.select_dtypes(include=['number']).columns

X_train[numeric_X] = imputer_X.fit_transform(X_train[numeric_X])
X_test[numeric_X] = imputer_X.transform(X_test[numeric_X])
```

```python
X_train.head()
```

```python
correlation = X[numeric_X].corr()
correlation
```

```python
categoric_X = X.select_dtypes(include=['category','object']).columns
```

```python
from sklearn.preprocessing import OneHotEncoder

one = OneHotEncoder(sparse_output=False)
```

```python
X_train_one_hote = one.fit_transform(X_train[categoric_X])
X_test_one_hote= one.transform(X_test[categoric_X])
 
```

```python
encoded_columns = one.get_feature_names_out(categoric_X)
```

```python
X_train_one_hote = pd.DataFrame(X_train_one_hote,columns=encoded_columns,index=X_train.index)
X_test_one_hote = pd.DataFrame(X_test_one_hote,columns=encoded_columns,index=X_test.index)
```

```python
X_train = X_train.drop(columns=categoric_X).join(X_train_one_hote)
X_test = X_test.drop(columns=categoric_X).join(X_test_one_hote)
```

```python
X_train
```

```python
X_train.shape
```

```python
from sklearn import ensemble
import xgboost as xgb
import lightgbm as lgb

ada_boost_regressor = ensemble.AdaBoostRegressor()

extra_trees_regressor = ensemble.ExtraTreesRegressor()
gradient_boosting_regressor = ensemble.GradientBoostingRegressor()
hist_gradient_boosting_regressor = ensemble.HistGradientBoostingRegressor()
random_forest_regressor = ensemble.RandomForestRegressor()

stacking_regressor = ensemble.StackingRegressor(
    estimators=[
        ('gb', gradient_boosting_regressor),
        ('rf',random_forest_regressor),
    ]
)
bagging_regressor = ensemble.BaggingRegressor(
)

xgb_regressor = xgb.XGBRegressor()
lgb_regressor = lgb.LGBMRegressor()

models = {
    "ada_boost": ada_boost_regressor,
    "bagging": bagging_regressor,
    "extra_trees": extra_trees_regressor,
    "gradient_boosting": gradient_boosting_regressor,
    "hist_gradient_boosting": hist_gradient_boosting_regressor,
    "random_forest": random_forest_regressor,
    "stacking": stacking_regressor,
    "xgboost": xgb_regressor,
    "lightgbm": lgb_regressor
}
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

results = {}

for name, model in models.items():

    trained_model = model.fit(X_train, y_train)
    y_preds = trained_model.predict(X_test)

    r2_n = r2_score(y_test, y_preds)
    mae = mean_absolute_error(y_test, y_preds)
    mse = mean_squared_error(y_test, y_preds)
    rmse = np.sqrt(mse)

    results[name] = {
        "R2": r2_n,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse
    }
```

```python
results = pd.DataFrame(results)
```

# Random Forest Regressor

```python
import pandas as pd

df = pd.read_csv('../job_salary_prediction_dataset.csv')
```

```python
X = df.drop('salary',axis = 1)
y = df['salary']
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(
  X,
  y,
  random_state=42,
  test_size=0.2
)
```

```python
from sklearn.ensemble import RandomForestRegressor

randomforest_reg = RandomForestRegressor(
  random_state=42,
  n_estimators=100,
  oob_score=True,
  bootstrap=True,
  max_depth=20
)
```

# We used oob_score to track generalization performance. If the OOB score does not improve with additional trees, we can interpret this as a saturation point and stop increasing n_estimators to avoid unnecessary model complexity.

```python
from sklearn.preprocessing import OneHotEncoder

one_hot = OneHotEncoder(sparse_output=False)
```

```python
categoric_data = X.select_dtypes(include=['category','object']).columns
```

```python
X_train_one_hot = one_hot.fit_transform(X_train[categoric_data])
X_test_one_hot = one_hot.transform(X_test[categoric_data])
```

```python
encoded_features = one_hot.get_feature_names_out()
```

```python
X_train_one_hot = pd.DataFrame(X_train_one_hot,columns=encoded_features,index=X_train.index)
X_test_one_hot = pd.DataFrame(X_test_one_hot,columns=encoded_features,index=X_test.index)
```

```python
X_train = X_train.drop(columns=categoric_data).join(X_train_one_hot)
X_test = X_test.drop(columns=categoric_data).join(X_test_one_hot)
```

```python
model = randomforest_reg.fit(
  X_train,
  y_train
)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
import numpy as np

y_preds = model.predict(X_test)

random_r2 = r2_score(y_test,y_preds)
random_mae = mean_absolute_error(y_test,y_preds)
random_mse = mean_squared_error(y_test,y_preds)
random_rmse = np.sqrt(mean_squared_error(y_test,y_preds))
print(random_r2,random_mae,random_mse,random_rmse)
```
