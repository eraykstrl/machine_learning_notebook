There is different selection and processing algorithm like OneHoteEncoding and Imputer.

One Hot Encoding converts categoric data to binary data it is useful for Deep Learning algorithms.
Imputing operiations fill missing values with mean, median etc.

```python
from sklearn.impute import SimpleImputer
import pandas as pd
```

```python
data = pd.read_csv("../job_salary_prediction_dataset.csv")
```

```python
data.head()
```

```python
df_categoric = data.select_dtypes(include = ['object','category','string'])
```

```python
df_categoric.head()
```

```python
df_numeric = data.select_dtypes(include = ['number'])
```

```python
df_numeric.head()
```

```python
df_numeric.isnull().sum()
```

```python
import numpy as np

imputer = SimpleImputer(missing_values = np.nan,strategy = "mean")
df_numeric = imputer.fit_transform(df_numeric)
```

```python
df_categoric = pd.get_dummies(df_categoric)
```

```python
df_categoric.head()
```

```python
df_categoric_2 = data.select_dtypes(include = ['object','category','string'])
```

```python
from sklearn.preprocessing import OneHotEncoder

one_hot_encoder = OneHotEncoder(sparse_output = False)

df_categoric_2 = one_hot_encoder.fit_transform(df_categoric_2)
```

```python
df_categoric_2 = pd.DataFrame(df_categoric_2)
```
